### Vapor–Liquid Equilibrium (NRTL)

This cell calculates **bubble and dew points** for a selected mixture using NRTL activity coefficients and two vapor-phase approximations:

- **γ–ideal:** Ideal gas vapor (φ_i=1).
- **γ–φ:** Non-ideal vapor using the second virial coefficient correlation of Tsonopoulos.

#### Instructions

1. **Select components:** Define the mixture in `components`. The order determines the order of mole fractions in `x` and `y`.
2. **Set conditions:** Specify temperature `T` (K) or pressure `P` (bar), together with the known liquid composition `x` or vapor composition `y`.
3. **Choose a calculation:**
   - `Ty_bub`: Bubble temperature and vapor composition, given `P, x`.
   - `Tx_dew`: Dew temperature and liquid composition, given `P, y`.
   - `Py_bub`: Bubble pressure and vapor composition, given `T, x`.
   - `Px_dew`: Dew pressure and liquid composition, given `T, y`.
4. **Select vapor model:** Use the `_NRTL` functions for ideal vapor or `_NRTL_virial` for the γ–φ approach.

All functions return the calculated temperature or pressure and the corresponding equilibrium composition. Mole fractions must follow the component order and sum to one.

**Units:** Temperature in K, pressure in bar, and dimensionless mole fractions.


```python
from properties import get_mixture
from equilibrium import Ty_bub_NRTL, Tx_dew_NRTL, Py_bub_NRTL, Px_dew_NRTL, Ty_bub_NRTL_virial, Tx_dew_NRTL_virial, Py_bub_NRTL_virial, Px_dew_NRTL_virial

#################
# STEP 1: Load parameters
#################

# ------ Load components here: ------------- 
# Example for IPA + Water:              components = ["IPA", "Water"]
# Example for Acetone + Water:          components = ["Acetone", "Water"] 
# Example for IPA + Acetone + Water     components = ["IPA", "Acetone", "Water"]
components = ["IPA", "Water"]           # <--------------------- Change this
fn_psats, nrtl_params, virial_params = get_mixture(components)


###################
# Step 2: Define (P (bar) or T (K)) and (x or y)
# x and y are define in the same order as defined in the "component" variable of STEP 1
###################

P = 1.01325 # bar       # <--------------------- Change this
x = [0.1, 0.9]          # <--------------------- Change this


################
# STEP 3: Choose model and calcule
# List of available functions:
# For gamma - ideal gas: Ty_bub_NRTL, Tx_dew_NRTL, Py_bub_NRTL, Px_dew_NRTL
# For gamma - phi: Ty_bub_NRTL_virial, Tx_dew_NRTL_virial, Py_bub_NRTL_virial, Px_dew_NRTL_virial
################

# Example gamma - ideal:
T1, y1 = Ty_bub_NRTL(P, x, fn_psats, nrtl_params)       # <--------------------- Change this
print(f"Result gamma - ideal: T={T1:.2f}, y1={y1}")

# Example gamma - virial:
T2, y2 = Ty_bub_NRTL_virial(P, x, fn_psats, nrtl_params, virial_params) # <--------------------- Change this
print(f"Result gamma - phi: T={T2:.2f}, y1={y2}")

# Its the same logic for P and for x, for instance:
# P1, y1 = Py_bub_NRTL(T, x. fn_psats, nrtl_params)
```
